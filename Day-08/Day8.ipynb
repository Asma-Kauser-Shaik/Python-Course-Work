# loops

### while 

```python
# epeatedly executes a block of code as long as the given condition remains true. When the condition becomes false, the line immediately after the loop in the program is executed.
```

```python
# while loop 
i = 1
while (i<= 10):
    print(i)
    i += 1
```

```python
# Example-2 : break statement
# break statement helps to come out of the loop
i = 1
while i <= 10:
     print(i)
     if i == 5:
         break
     i += 1    
```

```python
# Example-3: continue statement
# continue will take the control to the start of the loop
i =1
while i<10:
    i+= 1
    if i<5:
        continue
    print(i)    
```

```python
# Example-4 : else statement with while loop
# else statement executes only if loop is successfully completed

i = 1
while i<=10:
    print(i)
    i+=1
    if i > 5:
        break
else:
    print("loop is successfully completed")
```

```python
# Example- 5 : pass statement
# pass statement is a placeholder for future code

i = 0
while i<=10:
    if i==5:
        pass
    else:
        print(i)
    i+=1
```
