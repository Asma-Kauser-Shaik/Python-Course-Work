### Insert in set(add(),update())

```python
set1 = {10,20.5,"hello"}
# add() function is used to insert one item in the set
set1.add("world")
print(set1)
```

```python
set1.add((10,20,30,40))
print(set1)
```

```python
# update () is used for inserting multiple items together
set1.update((1000,200))
print(set1)
```

```python
set1.update({(1000,2000)})
print(set1)
```

# delete opeartion on set (remove(),discard(),pop(),clear(),del())

```python
dinner_set = {"Bagara rice","Biryani","pappu","chapati","fish","Mutton","prawns"}
# remove() function delete the one item whose name is provided
dinner_set.remove("prawns")
# discard() function delete the one item whose name is provided
dinner_set.discard("Biryani")
print(dinner_set)
```

```python
# discard Donot rasie ERROR if item is not present
dinner_set.discard("Biryani")
print(dinner_set)
```

```python
# pop is delete any random item from the  set
dinner_set.pop()
print(dinner_set)
```

```python
# clear() 
dinner_set.clear()
print(dinner_set)
```

```python
del dinner_set()
```
