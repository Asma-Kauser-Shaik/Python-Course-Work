# for loop

```python
# For loops is used to iterate over a sequence such as a list, tuple, string or range. It executes a block of code once for each item in the sequence.
```

```python
animals = ["puli","Pili","kukka","pandi","kothi"]
for i in animals:
    print(i)
```

```python
food_name = "Chicken_Dum_Masala_Biryani"
for i in food_name:
    print(i)
```

```python
# for Loop with range()
# range function is used to generate list of elements
# syntax of range : range(start,end,step)
# start = 0
# end = end -1
for i in range(10):
    print(i)
```

```python
# for loop with range()
# range() function is used to generate list of elements 
# syntax of range : range(start, end, step)
# start = 0
# end = end - 1
# step = 1
for i in range (4,10):
    print(i)
```

```python
# Example-4: break statement
# break statement will bring the control out of the loop

for num in range(1,6):
    if num == 3:
        break
    print(num)# for loop with range()
# range() function is used to generate list of elements 
# syntax of range : range(start, end, step)
# start = 0
# end = end - 1
# step = 1
for i in range (4,10,2):
    print(i)
```

```python
# Example-4: break statement
# break statement will bring the control out of the loop

for num in range(1,6):
    if num == 3:
        break
    print(num)
```

```python
# Example-5: Continue statement
# continue statement take the control to the start of the loop

for num in range(1,6):
    if num == 3:
        continue
    print(num)
```

```python
# Nested Loops
#A nested loop is a loop inside another loop. The inner loop executes completely for every iteration of the outer loop.
```

```python
# Example-6 : nested loop
# for every value of outer loop, inner loop will completely execute
for i in range(3,6): #n = 3 times 
    for j in range(2,7): # m = 5 times
        print(i,j)
```

```python
# Example-7 : else statement
# else statement executes if loop complete execution successfully

for i in range(5):
    if i == 2:
        continue
    elif i == 40:
        break
    else:
        print(i)
    pass
else:
    print("loop completed successfully")
```

# patterns

```python
#  square pattern
for j in range(5):
    for i in range(5):
        print("*", end = " ")
    print()
```

```python
# Right Triangle
for j in range(1,6):
    for i in range(j):
        print("*", end = " ")
    print()
```

```python
# inverted triangle
for j in range(5,0,-1):
    for i in range(j):
        print("*", end = " ")
    print()
```

```python
# Triangle
n = 5
for i in range(n, 0, -1):
    print(" " * i + "* " * (n-i+1))
```
