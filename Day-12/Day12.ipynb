```python
# convert list and set to tuple using tuple()
list1 = [10, 20, 30, 40]
tuple1 = tuple(list1)
print(tuple1)
```

```python
set1 = {10, 20, 30, 40}
tuple1 = tuple(set1)
print(tuple1)
```

# Insert, delete and update operations indirectly

```python
tuple1 = (10, 20.5, "Hello", [10, 20, 30])
tuple1[3][2] = 5000
```

```python
tuple1
```

```python
# tuple items cannot be updated directly
# complete item in tuple cannot be updated directly
# but if item is mutable then elements inside this mutable item can be updated
```

```python
#insert
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1.extend([100, 200, 300])
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
#delete
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1.remove(30)
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
# update
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1[2] = 300000
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

# popular inbuilt function for tuple

```python
Salary = (6, 70, 100, 600, 700, 90, 10, 80, 120)

print(tuple(reversed(Salary)))
print(len(Salary))
print(min(Salary))
print(max(Salary))
print(sum(Salary))

sorted_list = tuple(sorted(Salary, reverse = True))
print(sorted_list)

# sort() is inplace function
# inplace means sorting is done in the same data sturcture

# sorted() function return the list
# sort() function don't return anything
```

```python
l1 = [34, 40, 50, 60, 1, 3, 4, 70]
l1.sort()
print(l1)
```
