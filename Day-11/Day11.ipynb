# Updating list items

```python
list_politician = ["Vijay", "CBN", "Pavan", "JaiLalita", "Balayya", "Anushka"]

# update single element
list_politician[1] = "Dance with"

print(list_politician)
```

```python
# update multiple elements: update more elements with less
list_politician[3:5] = ["Dance with"]
print(list_politician)
```

```python
# update less elements with more
list_politician[1:4] = ["Rahul", "Kejriwal", "Yogi", "Mamta"]
print(list_politician)
```

```python
list_politician[-2] = "Dance with"
print(list_politician)
```

```python
list_politician[-5:-1] = ["Dance with"]
print(list_politician)
```

# insert items in list

```python
# append(), extend(), insert()
```

```python
l2 = []

# append() function is used to insert the single element at the last of the list
l2.append(10)
print(l2)
```

```python
l2.append(20)
print(l2)
```

```python
l2.append(30)
print(l2)
```

```python
# extend() : this function is used to insert multiple elements in the last

l2.extend([40, 50, 60, 70, 80])
print(l2)
```

```python
# insert() can be used to insert at specific index
l2.insert(4, "Dinesh")
print(l2)
```

```python
l2.insert(5, [10, 20, 30])
print(l2)
```

# delete item from list (remove(), pop(), clear(), del)

```python
food_list = [ "Papu", "Biryani", "fried rice", "Sambar", "Payasum", "Curd rice"]

# remove(): delete the item with given name
food_list.remove("fried rice")
print(food_list)
```

```python
# pop() is used to delete the item from a particular index
food_list.pop(4)
print(food_list)
```

```python
# pop() without index will delete the last element
food_list.pop()
print(food_list)
```

```python
food_list = [ "Papu", "Biryani", "fried rice", "Sambar", "Payasum", "Curd rice"]

# delete multiple elements together
del food_list[1:4]

print(food_list)
```

```python
# clear() is used to delete all the items from the list
food_list.clear()
print(food_list)
```

```python
# delete the list iteself
del food_list
```

# popular list inbuilt functions

# list comprehension

```python
# creating the list in a single line using a loop

l1 = [i for i in range(1,11)]
print(l1)
```

```python
l2 = [i**2 for i in range(1,11)]
print(l2)
```

```python
l3 = [i for i in range(1,11) if i%2==0]
print(l3)
```

```python
for i in range(1,11):
    if i%2==0:
        print(i)
```

```python
l3 = [i for i in range(1,11) if i%2==0 if i%3==0]
print(l3)
```

```python
l4 = [i**2 if i%2==0 else i+2 for i in range(1,11)]
print(l4)
```
