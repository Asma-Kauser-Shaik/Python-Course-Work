# Type casting

```python
# Type casting is the process of converting a value from one data type to another data type.
```

```python
Common Type Casting Functions
Function	Converts to	Example
int()	Integer	int("10")
float()	Float	float("10.5")
str()	String	str(100)
bool()	Boolean	bool(1)
list()	List	list("ABC")
tuple()	Tuple	tuple("ABC")
```

```python
# integer typecasting: int()
v1 = int(5.6)
v2 = int("300")
v3 = int(True)
v4 = int(False)
v5 = int("101",2)
print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
print(v5,type(v5))
```

```python
# float typecasting
v1 = float(56)
v2 = float("300")
v3 = float(True)
v4 = float(False)
print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
```

```python
# string str()
v1 = str(56)
v2 = str(30.0)
v3 = str(True)
v4 = str(False)
print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
```

```python
# boolean : bool()
v1 = bool(1)
v2 = bool(0)
v3 = bool(1000)
v4 = bool("Hello")
print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
```
