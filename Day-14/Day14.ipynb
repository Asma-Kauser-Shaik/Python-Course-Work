# Dictionary data structure

### different ways to create dictionary

```python
empty_dict = {}
empty_dict2 = dict()

print(empty_dict2)
print(type(empty_dict2))

int_dict = {1: 'Mamidipandu' , 2 : 'Aratipandu', 3: 'Batai'}
print(int_dict)

float_dict = {1.2: 'Mamidipandu' , 2.5 : 'Aratipandu', 3.7: 'Batai'}
print(float_dict)

str_dict = {"A": 'Mamidipandu' , "B" : 'Aratipandu', "C": 'Batai'}
print(str_dict)

het_dict = {10: 'Mamidipandu' , 20.5 : 'Aratipandu', "C": 'Batai'}
print(het_dict)

tuple_dict = {(10,20): {100,200} , 20.5 : 'Aratipandu', "C": 'Batai'}
print(tuple_dict)

# tuple can be used as a key
# list and set cannot be used as a key
# keys must be immutable

# values can be mutable and immutable

dict_comp = {x : x**2 for x in range(1,6)}
print(dict_comp)
```

# Accessing dictionary items

```python
drink_items = { 
                "Thumpsup" : 50,
                "Sugarcane" : 20,
                "CoconutWater" : 60,
                "Water" : 20,
                "rosemilk" : 30
}

val = drink_items["rosemilk"]
print(val)
```

```python
val2 = drink_items.get("rosemilk", "key not available")
print(val2)
```

```python
for i in drink_items:
    print(i, drink_items[i])
```

```python
print(drink_items.keys())
```

```python
print(drink_items.values())
```

```python
print(drink_items.items())
```

```python
# nested dictionary
nested_dict = {
                100 : {"Hello" : 10, "World" : 20},
                200 : {"A" : 50, "B" : 60}
}

nested_dict[100]["World"]
```

# insert & update operations on dictionary

```python
x_dict = {"name" : "Chinki", "age" : 80}

x_dict["age"] = 180    # update
print(x_dict)

x_dict['city'] = "Kurnool"  # insert
print(x_dict)

# insert and update using the update() function
# update() can insert and update multiple items together
x_dict.update({'age' : 18, 'occupation' : 'IAS', "name" : "minki", "Salary" : 1.5})
print(x_dict)
```

# delete the items from dictionary

```python
# (pop(), popitem(), clear(), del)

x_dict
```

```python
# pop() removes the item whose key is specified
x_dict.pop('age')
print(x_dict)
```

```python
# popitem() will delete the last element
x_dict.popitem()
print(x_dict)
```

```python
x_dict.clear()
print(x_dict)
```
