# Functions

```python
# Def:  A function is a block of reusable code that perform a specific task.

# define the function
def fun():
    print("Welcome to functions")
    print("Hello world")

fun()  # calling the function
fun()
```

```python
# Example-2: passing parameter in the function

# define the function
def fun(val1, val2):
    print("Welcome to functions", val1)
    print("Hello world", val2)

fun(100, 200)  # calling the function
```

```python
# example-3: nested function : function inside function

def Building():
    print("We are here")
    def Room1():
        print("Pavan: I am Here")
    Room1()  # calling from inside the Building function
    
Building()
```

```python
# example-4: Return statement 

def fun(a,b):
    c = a + b
    return c    # single value

val = fun(10, 20)
print(val)
```

```python
def get_name_age():
    return "Chinki", 125  # Multiple value

name, age = get_name_age()
print(name, age)
```

```python
def ds_fun():
    list1 = [10, 20, 30, 40]
    tuple1 = (50, 60, 70, 80)
    return list1, tuple1    # returing data sturcture

print(ds_fun())
```

```python
# return a function 
def outer():
    def inner():
        print("I am inner function")
    return inner   # return a function: return its starting address
    
f = outer()   # f is storing the reference of inner function
f()
```

# Default parameters

```python
# assign a default value to the "var" variable
def fun(var = "Nothing"):
    print(var)

fun()
fun(100)
```

```python
def fun2(x = 10, y = 20):
    print(x + y)

fun2()  # output = 30
fun2(100)  # output = 120
fun2(100, 200)
```

# passing arbitrary number of arguments using *args

```python
def pavan(**GF):
    for key in GF:
        print(key, " : " , GF[key])

pavan(Chinki=150, Minki=250)
pavan(Chinki=150)
pavan()
```

# lambda function

```python
obj = lambda a : a + 100

print(obj(10))
```

```python
obj = lambda a,b : a + b

print(obj(100, 200))
```

```python
obj1 = lambda a,b,c : a*b*c

val = obj1(1,2,3)
print(val)
```

```python
# lambda function are popularaly used inside the other function
# to perform some small task.

def fun(p):
    return lambda a : a**p

obj1 = fun(10)

print(obj1(2))
print(obj1(3))
print(obj1(4))
```

```python
# using lambda function inside the other function
def fun(p):
    obj = lambda a : a**p if a%2==0 else a+p
    return obj

obj = fun(2)

print(obj(2))
print(obj(3))
print(obj(4))
```
