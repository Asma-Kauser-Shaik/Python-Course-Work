# Input in python

```python
# input() is a built-in Python function used to take input from the user through the keyboard.
```

```python
# input 
var = input("Enter total no of students")
print(var)
```

```python
var = int(input("Enter total no of students"))
print(var)          
```

```python
var = float(input("Enter student marks"))
print(var)
```

```python
# split() is used to separte
x1 , x2,x3 = input("Enter 3 names").split()
print(x1,x2,x3)
```

```python
# map  convert the value into integer.
a ,b = map(int,input("Enter two numbers ").split())
print(a)
print(b)
print(a + b)
```

```python
x = list(map(int, input("Enter any number of values").split()))
print(x)
```

# Output in python


```python
# print() function allows us to display text, variables and expressions on the console. In the below example, "Hello, World!" is a string literal enclosed within double quotes. When executed, this statement will output the text to the console.
```

```python
# 1. concatenation using commas
# 2. String concatenation using +
# 3. str.format() method
# 4. f-strings method (python 3.6+ version)
# 5. Escape sequences
# 6. End and Sep parametes in print()
```

```python
 # concatentation using commas.
Name = "Asma"
age = 22
marks = 80.5
print(Name,age,marks)
```

```python
# String concatenation using + 
First_Name = "Asma"
Last_Name = "Kauser"
print(First_Name + " " + Last_Name + " ")
```

```python
# str format 
Name = "Asma"
Age = 22
print("Name: {0} Age: {1}" . format (Name , Age))
```

```python
# f strings 
first_name = "asma"
last_name = "kauser"
print(f"Name of the students : {first_name} ,{last_name}")
```

```python
First_Name = "Asma"
Last_Name = "Kauser"
print(f"First_Name : {First_Name} , \tLast_Name: {Last_Name}")
```

```python
First_Name = "Asma"
Last_Name = "Kauser"
print(f"First_Name : {First_Name} , \nLast_Name: {Last_Name}")
```

```python
First_Name = "Asma"
Last_Name = "Kauser"
print(f"First_Name : \"{First_Name} \", \nLast_Name: {Last_Name}")
```

```python
# end and sep parameters in print
# end = used to decided how the print statements works in end 
# sep how multiple values are to separated.
```

```python
Name = "Asma"
age = 22
marks = 80.5
print(Name,end = " ")
print(age,end = " ")
print(marks)
```

```python
Name = "Asma"
age = 22
marks = 80.5
print(Name, age , marks , sep = "@")
```
